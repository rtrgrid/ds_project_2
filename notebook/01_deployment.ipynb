# 🚀 Task 1: ONNX Export & CoreML Deployment

## 🎯 Objective
The goal of this task is to convert the trained Transformer model into a deployment-ready format and evaluate its performance on edge devices.

We aim to:
- Export PyTorch model → ONNX
- Validate numerical correctness
- Convert ONNX → CoreML
- Benchmark latency and memory

## 🧠 Background

PyTorch models are not directly usable in production environments like mobile or edge devices.

To deploy efficiently:
- ONNX provides a cross-framework representation
- CoreML enables optimized execution on Apple Silicon (M1/M2)

This pipeline bridges research → production.

## ⚙️ Step 1: Load Trained Model

We load the trained Transformer model and set it to evaluation mode to disable dropout and gradients.

## ⚙️ Step 2: Export to ONNX

We export the model using `torch.onnx.export`.

### ❌ Issues Faced

1. **Missing dependency**
   - Error: `ModuleNotFoundError: onnxscript`
   - Cause: New PyTorch exporter requires `onnxscript`

2. **Opset mismatch**
   - Error related to `LayerNormalization`
   - Cause: Using outdated opset (14)

---

### ✅ Fixes Applied

- Installed dependencies:
  - onnx
  - onnxruntime
  - onnxscript

- Used opset version ≥ 18 to support Transformer layers

---

### ⚠️ Important Insight

Transformer models rely on modern operators → older ONNX versions fail.

## 🔍 Step 3: Validate ONNX Model

To ensure correctness, we compare ONNX output with PyTorch output.

### Method:
- Run both models on same input
- Compute max absolute difference

---

### 📊 Result

Difference observed:
~2e-07

### ✅ Interpretation

- Outputs are nearly identical
- Confirms successful conversion

## 🍏 Step 4: Convert to CoreML

CoreML is optimized for Apple hardware (M-series chips).

---

### ❌ Issues Faced

1. **Static shape constraint**
   - CoreML does not support dynamic batch sizes

2. **Shape mismatch errors**
   - PyTorch supports batch (N, 24, 3)
   - CoreML expects fixed shape

---

### ✅ Fixes Applied

- Fixed input shape to:
  (1, 24, 3)

- Used TorchScript tracing before conversion

---

### ⚠️ Insight

Deployment frameworks often require stricter constraints than training frameworks.

## ⚡ Step 5: Latency Benchmarking

We measure inference time for:

- PyTorch
- CoreML

Across different batch sizes.

---

### 📊 Observations

- CoreML provides faster inference due to:
  - hardware optimization
  - float16 precision

- Batch size impacts latency:
  - small batch → low latency
  - large batch → higher throughput

---

### 🧠 Insight

Edge deployment benefits from:
- lower precision (float16)
- optimized runtimes (CoreML)

## 💾 Step 6: Memory Profiling

We measured memory usage at:
- model load
- inference

---

### 📊 Observations

- Slight increase after model load
- Stable memory during inference

---

### 🧠 Insight

Efficient memory usage is critical for real-time systems.

## 🏁 Summary

### ✅ Achievements

- Successfully exported model to ONNX
- Validated numerical correctness
- Converted to CoreML for edge deployment
- Benchmarked latency and memory

---

### 🧠 Key Learnings

- Deployment requires handling compatibility issues
- ONNX acts as a bridge between frameworks
- CoreML enables optimized edge inference
- Model correctness must always be validated after conversion

---

### 🚀 Conclusion

This task transformed the model from a research prototype into a deployable system suitable for real-world applications.
