# 🧪 Task 4: Positional Encoding Ablation Study

## 🎯 Objective
The goal of this task is to analyze the impact of positional encoding on Transformer performance for time-series forecasting.

---

We evaluate:
- Does positional encoding help?
- Which type performs best?
- Can the model generalize to longer sequences?

---

This task focuses on **understanding model behavior**, not just improving performance.

## 🧠 Background

Transformers do not inherently understand sequence order.

To encode positional information, we use:

- Sinusoidal encoding (fixed)
- Learnable embeddings
- Relative encoding (ALiBi)

---

### ❗ Why Important?

In time-series:
- Order of data is critical
- Model must know "when" events occurred

---

### ⚠️ Hypothesis

Different positional encodings may affect:
- accuracy
- generalization
- extrapolation ability

## ⚙️ Positional Encoding Variants

We tested four configurations:

---

### 1️⃣ Sinusoidal Encoding
- Fixed mathematical encoding
- No learning required
- Generalizes well

---

### 2️⃣ Learnable Embeddings
- Model learns position representation
- Flexible but limited to training range

---

### 3️⃣ ALiBi (Attention with Linear Bias)
- Adds bias based on distance
- Designed for long-range generalization

---

### 4️⃣ No Positional Encoding
- Model relies purely on values

---

### 🎯 Goal

Compare performance across these variants

```python
if len(X_test) == 0:
    continue



For each encoding type:

1. Initialize model with selected encoding
2. Run inference on test data
3. Compute RMSE

---

### 📊 Evaluation Metric

RMSE (Root Mean Squared Error)

---

### 🎯 Goal

Compare performance under identical conditions

## 📊 Results

| Encoding Type | RMSE |
|--------------|------|
| None         | **Best** |
| Sinusoidal   | Good |
| ALiBi        | Moderate |
| Learnable    | Poor |

---

### 🔍 Observations

- Learnable encoding performs worst
- Sinusoidal provides stable performance
- No encoding surprisingly performs best

## ❌ Challenge 2: Learnable Encoding Failure

### Observation:
Learnable embeddings gave very high RMSE

---

### 🧠 Root Cause

- Model learns positions only for training window (24 steps)
- Cannot generalize to unseen positions

---

### 💡 Insight

Learnable embeddings:
- Good for fixed-length tasks
- Poor for extrapolation

## 🤯 Why Did "No Positional Encoding" Perform Best?

### Possible Reasons:

1. Time-series values already encode temporal patterns
2. Data has strong local dependencies
3. Model relies on value trends rather than position

---

### 🧠 Interpretation

For this dataset:
→ positional information may be redundant

---

### 💡 Insight

More complex models are not always better

## 📏 ALiBi Analysis

ALiBi introduces a linear bias based on distance.

---

### 📊 Behavior

- Encourages attention to closer timesteps
- Supports longer context

---

### ⚠️ Observation

- Did not outperform sinusoidal
- Still better than learnable embeddings

---

### 💡 Insight

ALiBi is useful for long sequences, but dataset here is short

## 🔮 Extrapolation Analysis

We tested model on longer context than training.

---

### 📊 Findings

- Sinusoidal → stable
- ALiBi → reasonable
- Learnable → fails

---

### 🧠 Insight

Generalization depends on encoding type:
- Fixed encodings generalize better
- Learned encodings overfit

## 🏁 Summary

### ✅ Achievements

- Compared 4 positional encoding strategies
- Evaluated performance using RMSE
- Tested generalization to longer sequences

---

### 🧠 Key Learnings

- Positional encoding is not always necessary
- Learnable embeddings fail to extrapolate
- Sinusoidal and ALiBi are more robust
- Simpler models can outperform complex ones

---

### 🚀 Conclusion

This study shows that model design choices must be aligned with data characteristics, and positional encoding should not be blindly applied.
