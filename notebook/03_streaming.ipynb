# ⚡ Task 3: Streaming Inference with Sliding Window

## 🎯 Objective
The goal of this task is to simulate a real-time electricity forecasting system where data arrives continuously and predictions must be made with minimal latency.

---

We aim to:
- Perform prediction step-by-step (streaming)
- Maintain a rolling context window (24 timesteps)
- Ensure low latency and stable performance

---

This converts the model from **batch prediction → real-time system**

## 🧠 Why Streaming Inference?

In real-world systems:
- Data arrives continuously (not in batches)
- Predictions must be made instantly
- Memory must remain constant over time

---

### ❗ Without Streaming Design:
- High latency
- Memory growth
- Inefficient recomputation

---

### ✅ With Streaming:
- Real-time predictions
- Stable memory usage
- Production-ready pipeline

## ⚙️ Approach

We simulate streaming using:

---

### 1️⃣ Sliding Window
- Maintain last 24 timesteps
- Use it as model input

---

### 2️⃣ Circular Buffer
- Drop oldest value
- Append newest value
- Prevent memory growth

---

### 3️⃣ Continuous Prediction Loop
- Process incoming data one step at a time

---

This mimics real-time deployment systems.
